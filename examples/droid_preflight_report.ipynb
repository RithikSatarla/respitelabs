# Kintrace on one DROID episode (CPU)

Runs `kintrace preflight` and `kintrace report` on one real DROID episode.
Needs the local DROID data (`python -m kintrace.droid download ...`, then `fklabels`) and the detector at `data/droid/detector_final.pt`.
Everything here runs on the CPU.

```python
import json, os, subprocess, sys
os.environ['CUDA_VISIBLE_DEVICES'] = ''   # CPU only
REPO = os.path.abspath('..') if os.path.basename(os.getcwd()) == 'examples' else os.getcwd()
os.chdir(REPO)
man = {m['key']: m['dir'] for m in json.load(open('data/droid/manifest.json'))}
EPISODE = man['TRI+52ca9b6a+2023-11-02-16h-42m-30s']    # a healthy test episode
BASELINE = man['TRI+52ca9b6a+2023-11-01-11h-00m-06s']   # an earlier healthy session, same camera
CAMERA = '20252535'
print(EPISODE)
```

## Preflight
Every check that applies, then GO / NO-GO. Exit code 0 is GO, 1 is NO-GO, 2 is can't tell.

```python
r = subprocess.run([sys.executable, '-m', 'kintrace', 'preflight', EPISODE, '--camera', CAMERA, '--baseline', BASELINE],
                   capture_output=True, text=True)
print(r.stdout)
print('exit code', r.returncode)
```

## Report
One self-contained HTML page: expected vs detected keypoints, what moved with error bars, the fix, before vs after.

```python
out = 'data/droid/reports/notebook_report.html'
r = subprocess.run([sys.executable, '-m', 'kintrace', 'report', EPISODE, '--camera', CAMERA, '--out', out],
                   capture_output=True, text=True)
print(r.stdout or r.stderr[-2000:])
from IPython.display import IFrame
IFrame(out, width='100%', height=900)
```
